### Basic ploting 

```python
import matplotlib.pyplot as plt
x=[82,32,45,6,78]
y=[6,4,54,67,8]

plt.plot(x,y)
plt.xlabel("time")
plt.ylabel("didtance")
plt.title("Time vs distance graph")
plt.show()
```

### Malti line ploting

```python
import matplotlib.pyplot as plt
x=[82,32,45,6,78]
y=[6,4,54,67,8]
z=[82,32,45,6,78]
plt.plot(x,y , label="Y function")
plt.plot(x,z ,label="Z function")
plt.xlabel("time")
plt.ylabel("didtance")
plt.title("Time vs distance graph")
plt.legend()
plt.show()
```

### Sub ploting


```python
fig, axes = plt.subplots(1, 3, figsize=(10, 4))
x=[82,32,45,6,78]
y1=[6,4,54,67,8]
y2=[82,32,45,6,78]
axes[0].plot(x, y1)
axes[0].set_title("Plot 1")
axes[1].plot(x, y2)
axes[1].set_title("Plot 2")
axes[2].plot(x,x)
axes[2].set_title("Plot 3")
plt.tight_layout()
plt.show()
```

### Image display

```python
import numpy as np
from skimage import data

image = data.camera()  # Load sample image
plt.imshow(image, cmap='gray')
plt.title("Grayscale Image")
plt.axis('on')  # Hide axes
plt.show()
```

### Scattering 

```python
x = np.random.randn(100)
y = np.random.randn(100)

plt.scatter(x, y, c='green', marker='o', s=20)
plt.title("Scatter Plot")
plt.show()
```

### Histogram

```python
data = np.random.randn(1000)

plt.hist(data, bins=10, edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Griding 

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, 'g', linewidth=2, label='sin(x)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
